In Klasse 10 geht es um das Verständnis von Atombilanzen. Eine Reaktion wie die Verbrennung von Propan (a⋅C3H8+b⋅O2→c⋅CO2+d⋅H2O) wird in einfache Bilanzen zerlegt:

```python
import sympy as sp

# 1. Unbekannte Stöchiometriefaktoren als mathematische Symbole definieren
a, b, c, d = sp.symbols("a b c d")

# 2. Elementbilanzen aufstellen (Anzahl Atome links = Anzahl Atome rechts)
# C-Bilanz: 3*a = c       -> 3*a - c = 0
# H-Bilanz: 8*a = 2*d     -> 8*a - 2*d = 0
# O-Bilanz: 2*b = 2*c + d -> 2*b - 2*c - d = 0

gleichungen = [3 * a - c, 8 * a - 2 * d, 2 * b - 2 * c - d]

# 3. System lösen in Abhängigkeit von 'a' (für a = 1 Molekül Propan)
loesung = sp.solve(gleichungen, (b, c, d))

print("⚖️ Ausgeglichene Faktoren für 1 Mol Propan (a = 1):")
print(f"a = 1  |  b = {loesung[b]}  |  c = {loesung[c]}  |  d = {loesung[d]}")
```

Klasse 11 (MSS): Redoxreaktionen & Ladungsbilanzen

In der Oberstufe kommen zu den Atombilanzen noch Ladungsbilanzen hinzu. Hier lässt sich das Gauß-Verfahren aus der linearen Algebra nutzen, um komplexe Redoxgleichungen (wie Permanganat mit Eisen(II)) mathematisch exakt zu lösen:

```python
import sympy as sp

# Reaktion: a MnO4- + b Fe2+ + c H+  ->  d Mn2+ + e Fe3+ + f H2O
# Zeilen der Matrix: [Mn, O, Fe, H, Ladung]
matrix = sp.Matrix(
    [
        [1, 0, 0, -1, 0, 0],  # Mangan-Bilanz
        [4, 0, 0, 0, 0, -1],  # Sauerstoff-Bilanz
        [0, 1, 0, 0, -1, 0],  # Eisen-Bilanz
        [0, 0, 1, 0, 0, -2],  # Wasserstoff-Bilanz
        [-1, 2, 1, -2, -3, 0],  # Ladungsbilanz
    ]
)

# Der Kern (Nullraum) der Matrix liefert direkt die Stöchiometrie
loesungsvektor = matrix.nullspace()[0]

# Umwandlung in ganze Zahlen
faktoren = loesungsvektor * sp.lcm([val.q for val in loesungsvektor])

print("⚖️ Berechnete Redox-Koeffizienten [a, b, c, d, e, f]:")
print(faktoren)
```

Lotka-Volterra-Simulation (Räuber-Beute-Dynamik)

```python
import matplotlib.pyplot as plt
import numpy as np

# --- Parameter der Simulation ---
dt = 0.01  # Zeitschritt
schritte = 2000

# Startpopulationen anlegen
beute = np.zeros(schritte)
raeuber = np.zeros(schritte)
beute[0] = 40  # Startanzahl Beute (z. B. Hasen)
raeuber[0] = 9  # Startanzahl Räuber (z. B. Füchse)

# Koeffizienten (Fortpflanzung, Sterblichkeit, Begegnung)
alpha = 1.0  # Beute-Geburtenrate
beta = 0.1  # Beute-Sterberate durch Räuber
gamma = 1.5  # Räuber-Sterberate
delta = 0.075  # Räuber-Zuwachs pro gefressener Beute

# --- Berechnung Schritt für Schritt ---
for t in range(schritte - 1):
    d_beute = (alpha * beute[t] - beta * beute[t] * raeuber[t]) * dt
    d_raeuber = (
        delta * beute[t] * raeuber[t] - gamma * raeuber[t]
    ) * dt

    beute[t + 1] = beute[t] + d_beute
    raeuber[t + 1] = raeuber[t] + d_raeuber

# --- Diagramm erstellen ---
zeit = np.arange(schritte) * dt
plt.figure(figsize=(9, 4))
plt.plot(zeit, beute, label="Beute 🐰", color="green", linewidth=2)
plt.plot(zeit, raeuber, label="Räuber 🦊", color="red", linewidth=2)

plt.title("📈 Lotka-Volterra-Modell: Räuber-Beute-Dynamik")
plt.xlabel("Zeit $t$")
plt.ylabel("Individuenzahl")
plt.legend()
plt.grid(True)
plt.show()
```
