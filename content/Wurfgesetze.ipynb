### 📐 Physikalische Grundlagen
Beim waagerechten Wurf überlagern sich zwei ungestörte Teilbewegungen (**Superpositionsprinzip**):

1. **In $x$-Richtung (horizontal):** Gleichförmige Bewegung mit konstanter Geschwindigkeit $v_0$
   $$x(t) = v_0 \cdot t$$

2. **In $y$-Richtung (vertikal):** Gleichmäßig beschleunigte Bewegung durch die Erdbeschleunigung $g \approx 9{,}81\text{ m/s}^2$
   $$y(t) = h - \frac{1}{2} \cdot g \cdot t^2$$

### 📝 Aufgabe 1: Flugdauer und Wurfweite berechnen

Die gesamte Flugzeit $t_{\text{ges}}$ wird durch den Zeitpunkt bestimmt, an dem der Körper den Boden ($y = 0$) erreicht:
$$t_{\text{ges}} = \sqrt{\frac{2 \cdot h}{g}}$$

Die maximale Wurfweite $w$ ergibt sich durch Einsetzen von $t_{\text{ges}}$ in die $x$-Gleichung:
$$w = x(t_{\text{ges}}) = v_0 \cdot t_{\text{ges}}$$

**Verändere im folgenden Code die Werte für `h` und `v0` und führe die Zelle aus!**

```python
# --- EINGABEWERTE (Hier experimentieren!) ---
h = 10.0      # Abwurfhöhe in Metern (m)
v0 = 15.0     # Anfangsgeschwindigkeit in m/s
g = 9.81      # Erdbeschleunigung in m/s^2

# --- BERECHNUNG ---
# 1. Gesamte Flugzeit t_ges
t_ges = np.sqrt((2 * h) / g)

# 2. Maximale Wurfweite w
w = v0 * t_ges

# --- AUSGABE DER ERGEBNISSE ---
print("=" * 35)
print(f"Abwurfhöhe h:        {h:.1f} m")
print(f"Geschwindigkeit v₀:  {v0:.1f} m/s")
print("-" * 35)
print(f"Gesamte Flugdauer:   {t_ges:.2f} Sekunden")
print(f"Maximale Wurfweite:  {w:.2f} Meter")
print("=" * 35)
```

### 📊 Aufgabe 2: Flugbahn (Trajektorie) visualisieren

Mit der folgenden Zelle lassen wir Python die Flugbahn des Objekts grafisch als Diagramm zeichnen.

```python
# Zeitarray von t = 0 bis t_ges mit 100 Datenpunkten erzeugen
t = np.linspace(0, t_ges, 100)

# Orte x(t) und y(t) berechnen
x = v0 * t
y = h - 0.5 * g * t**2

# --- DIAGRAMM ERSTELLEN ---
plt.figure(figsize=(9, 5))
plt.plot(x, y, color='#1f77b4', linewidth=2.5, label=f'v₀ = {v0} m/s, h = {h} m')

# Diagramm gestalten
plt.title('Flugbahn beim waagerechten Wurf', fontsize=14, fontweight='bold')
plt.xlabel('Horizontale Entfernung x (in m)', fontsize=12)
plt.ylabel('Höhe y (in m)', fontsize=12)
plt.xlim(0, max(w * 1.1, 1))
plt.ylim(0, max(h * 1.1, 1))
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(fontsize=11)

# Aufprallpunkt hervorheben
plt.plot(w, 0, marker='o', markersize=8, color='red', label='Aufprallpunkt')

plt.show()
```
