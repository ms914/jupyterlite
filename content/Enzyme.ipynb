## Teil 1: Die Dreidimensionale Struktur der DNA (PDB: 1BNA)

Die B-Form der DNA ist die häufigste Konformation in biologischen Zellen. Sie bildet eine **rechtsgängige Doppelhelix** mit gegenläufigen (antiparallelen) Einzelsträngen.

* **Zucker-Phosphat-Rückgrat:** Außen liegend, hydrophil.
* **Stickstoffbasen (A-T, G-C):** Innen liegend, hydrophob, über Wasserstoffbrücken verbunden.

```python
# --- DNA-VISUALISIERUNG (1BNA: B-DNA Dodecamer) ---
view_dna = py3Dmol.view(query='pdb:1BNA')

# Hintergrund & Darstellungsstil
view_dna.setBackgroundColor('white')

# Rückgrat als Cartoon / Bänder darstellen, Basen als Stäbchen (Sticks)
view_dna.setStyle({'cartoon': {'color': 'spectrum'}, 'stick': {'radius': 0.2}})

# Ansicht zentrieren
view_dna.zoomTo()
view_dna.show()
```

## Teil 2: Enzymstruktur & Aktives Zentrum (PDB: 1A2C)

Proteine bestehen aus Aminosäureketten, die sich zu komplexen 3D-Strukturen falten ($\alpha$-Helices und $\beta$-Faltblätter). In Enzymen bildet sich dadurch eine spezifische Tasche aus: das **aktive Zentrum**.

Im folgenden Modell sehen wir das Enzym **Thrombin** (ein wichtiges Enzym der Blutgerinnung).
* **Cartoon (bunt):** Das Rückgrat des Proteingerüsts.
* **Kugeln (Rot):** Ein Inhibitor/Molekül, das direkt im aktiven Zentrum bindet.

```python
# --- ENZYM-VISUALISIERUNG (1A2C: Thrombin mit Inhibitor) ---
view_enz = py3Dmol.view(query='pdb:1A2C')
view_enz.setBackgroundColor('#f8f9fa')

# 1. Proteingerüst als Cartoon darstellen (gefärbt nach Sekundärstruktur)
view_enz.setStyle({'cartoon': {'color': 'spectrum'}})

# 2. Aktives Zentrum / Gebundenes Molekül hervorheben (Als Kugeln)
view_enz.addStyle({'resn': 'IH2'}, {'sphere': {'color': 'red', 'scale': 0.8}})

# 3. Naheliegende Aminosäuren des Enzyms am aktiven Zentrum als Stäbchen zeigen
view_enz.addStyle({'within': {'distance': 6, 'sel': {'resn': 'IH2'}}}, 
                  {'stick': {'colorscheme': 'cyanCarbon'}})

view_enz.zoomTo()
view_enz.show()
```
